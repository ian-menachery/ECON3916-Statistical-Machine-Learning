```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import statsmodels.api as sm
import statsmodels.formula.api as smf

# Configure default visual settings (Phase 2 conventions)
sns.set_theme(style="whitegrid")
plt.rcParams['figure.figsize'] = (10, 6)


df = pd.read_csv('ResumeNames.csv')

df['call'] = np.where(df['call'] == 'yes', 1, 0)
```

```python
formula_base = 'call ~ C(ethnicity) + C(quality) + experience + C(city)'

model_base = smf.ols(formula=formula_base, data=df).fit(cov_type='HC1')
print("--- Baseline Linear Probability Model ---")
print(model_base.summary())
```


Model Interpretations

Ethnicity (Treatment Effect): Ceteris paribus, Possessing a Caucasian-sounding name on a resume increases the probability of receiving an interview callback by 3.20 percentage points compared to an African-American-sounding name. This result is highly statistically significant (p < 0.001), isolating the causal effect of perceived race in the interview screening process.

Experience: Ceteris paribus, Each additional year of work experience is associated with a 0.28 percentage point increase in the probability of a callback, which is statistically significant (p = 0.001).

City: Ceteris paribus, Resumes submitted in the Chicago labor market experience a 2.45 percentage point lower callback probability compared to the Boston market, statistically significant at the 1% level.



```python
formula_interact = 'call ~ C(ethnicity) * C(quality) + experience + C(city)'

model_interact = smf.ols(formula=formula_interact, data=df).fit(cov_type='HC1')
print("\n--- Interaction Model (Heterogeneous Effects) ---")
print(model_interact.summary())
```

Interaction Model Interpretation Testing Heterogeneity:
Ethnicity & Resume Quality: The interaction term coefficient is -0.0177, but with a p-value of 0.254, this effect is not statistically significant. Ceteris paribus, we cannot reject the null hypothesis that the interaction effect is zero. This implies that the callback gap between Caucasian and African-American names persists regardless of resume quality, and showing improving credentials does not significantly close the racial callback gap.


Omitted Variable Bias (OVB) Argument:
Because the data originates from a Randomized Controlled Trial, the primary treatment variable ethnicity was randomly assigned to the standardized resumes. Because of this the treatment is statistically independent of any unobserved applicant characteristics (like work ethic or interview skills). Therefore, our estimate of the treatment effect does not suffer from traditional Omitted Variable Bias regarding applicant traits.
